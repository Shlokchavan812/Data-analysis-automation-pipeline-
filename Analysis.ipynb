```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt  
import seaborn 
```

```python
df=pd.read_csv("C:\\Users\\Shlok\\Desktop\\Nuclear_launcher_codes\\Data anaylsis project\\student_performance_clean_reference (1).csv")
df
```

```python
def clean_data(df):
    df=df.drop_duplicates()
    cols=["study_hours_per_week","attendance_pct","previous_marks"]
    for c in cols:
        df[c]=pd.to_numeric(df[c],errors="coerce")
    df=df[(df.attendance_pct<=100)&
          (df.final_marks<=100)] 
    df=df.fillna(df.median(numeric_only=True))
    return df
```

```python
def transform_data(df):
    df["grade"]=pd.cut(df.final_marks,bins=[0,49,79,100],labels=["Needs Improvement","Average","Excellent"])
    df["att_group"]=pd.cut(df.attendance_pct,bins=[0,75,90,100],labels=["Low","Medium","High"])
    return df
```

```python
def anaylse_data(df):
    summary=df.groupby("att_group").agg(avg_study=("study_hours_per_week","mean"),
                                        avg_attendance=("attendance_pct","mean"),
                                        Students=("student_id","count"))
    corr=df.corr(numeric_only=True)
    return summary,corr
```

```python
def load_resuts(df,summary):
    df.to_csv("clean_students.csv",index=False)
    summary.to_csv=("Analysis_result.csv")
```

```python
def run_pipeline(df):
    df=clean_data(df)
    df=transform_data(df)
    summary,corr=anaylse_data(df)
    load_resuts(df,summary)
    return df,summary,corr
```

```python
run_pipeline(df)
```
